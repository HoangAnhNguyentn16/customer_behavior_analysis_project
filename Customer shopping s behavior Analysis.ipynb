```python
#Thêm file csv vào trong python
import pandas as pd
df = pd.read_csv('E:\Data Analysis\Project\Portfolio 1 Phân tích hàng vi khách hàng\customer_shopping_behavior.csv')
```

```python
#Thống kê tất cá các cột những thông số cơ bản
df.describe(include = 'all')
```

```python
#Kiểm tra những dòng có có null
df.isnull().sum()
#Review Rating     có        37 null
```

```python
#Thay thế những thông tin bị null bằng giá trị trung vị (median) của danh mục đó
#Sử dụng fillna()
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
#Kiểm tra lại bảng
df.isnull().sum()
```

```python
#Chuyển tên cột về dạng dẽ đọc và nhớ hơn
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns ={'purchase_amount_(usd)':'purchase_amount'})
```

```python
#Kiểm tra lại data
df.columns
```

```python
#Tạo thêm 1 cột để nhóm các ddoooj tuổi khách hàng lại
labels = ['Young', 'Adult', 'Middle-age', 'Senior']
#Chia dữ liệu thành 4 nhóm bằng nhau dựa trên phân bố dữ liệu và gán những nhãn bên trên theo thứ tự
df['age_group'] = pd.qcut(df['age'],  q = 4, labels = labels)
```

```python
#Kiểm tra dữ liệu cột đã tạo
df[['age', 'age_group']].head(10)
```

```python
#Tần suất mua hàng theo ngày
#Tạo 1 từ điển với bản đồ tần suất
frequency_maping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 month' : 90
}
#sử dụng hàm map để tạo cột mới với thông tin trong cột frequency_of_purchases và tahy thế chúng bằng số ngày
#Tương ứng trong từ điển
df['Tần_Suất_Mua_Hàng'] = df['frequency_of_purchases'].map(frequency_maping)
```

```python
#Kiểm tra dữ liệu đã tạo ra
df[['frequency_of_purchases', 'Tần_Suất_Mua_Hàng']].head(10)
```

```python
#Cột "Đã áp dụng giảm giá" và cột "Đã áp dụng mã giảm giá"
df[['discount_applied', 'promo_code_used']].head(15)
```

```python
#Kiểm tra xem thông tin 2 cột có giống nhau không 
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Vì là thông tin giống nhau, xóa bớt 1 cột đi
df = df.drop('promo_code_used', axis=1)
```
